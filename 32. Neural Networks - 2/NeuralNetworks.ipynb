```python
import torch
from torch import nn
#why Pytorch or TF because they let us use GPUs efficiently!
```

```python
#Tensors for Data Handling..!
X = torch.arange(0,11,dtype=torch.float32).reshape(-1,1)
```

```python
X
```

```python
y = torch.tensor([20,25,30,35,50,65,80,95,110,125,140], dtype = torch.float32).reshape(-1,1)
```

```python
y
```

```python
X.shape
```

```python
#nn.Module is allowing SalaryNet to inherit all nn modular functions from Torch
#By default, PyTorch initializes the weights and biases of basic layers (such as nn.Linear and nn.Conv2d) randomly from a uniform distribution.
class SalaryNet(nn.Module):
  def __init__(self):
    super().__init__()
    self.hidden = nn.Linear(in_features = 1, out_features = 2)#input is X output h1 and h2
    self.relu = nn.ReLU()
    self.output = nn.Linear(2,1)

  def forward(self,x):
    h = self.hidden(x) #weighted Sum Z
    h = self.relu(h) #Activation
    salary = self.output(h)
    return salary

model = SalaryNet()
```

```python
with torch.no_grad(): #not final trained parameters just a sensible starting point for this class
  model.hidden.weight.copy_(torch.tensor([[1.2],[0.8]]))
  model.hidden.bias.copy_(torch.tensor([0.4,-1.6]))
  model.output.weight.copy_(torch.tensor([[3.0,7.0]]))
  model.output.bias.fill_(torch.tensor(15.0))
```

```python
print(model)
```

```python
loss_fn = nn.MSELoss()
```

```python
optimizer = torch.optim.Adam(model.parameters(),lr=0.02) #Optimizers + different types of GDs next class
```

```python
#Each epoch: predict -> measure loss -> compute Gradients -> nudge the weights

for epoch in range(3000):
  prediction = model(X)
  loss = loss_fn(prediction,y)

  optimizer.zero_grad() #LIKE FACTORY RESET - Preparing my optimizer
  loss.backward()
  optimizer.step()

  if (epoch+1)%500 == 0:
    print(f"Epoch {epoch+1}, Loss:{loss.item():.6f}")
```

```python
model.eval()

with torch.no_grad():
  experience = torch.tensor([[5.0]])
  predicted_salary = model(experience).item() *1000

print(f"Predicted Salary for 5 years: Rs {predicted_salary:,.0f}")
```
