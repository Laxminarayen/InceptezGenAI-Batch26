```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
from scipy.signal import correlate2d
```

```python
image = data.camera()
```

```python
plt.imshow(image, cmap = 'gray')
```

```python
npimage
```

```python
image.shape
```

```python
filter_kernel = np.array([
    [-1,-1,-1],
    [0,0,0],
    [1,1,1]
])
```

```python
output_image = correlate2d(image, filter_kernel, mode = 'same')
```

```python
plt.imshow(np.maximum(output_image,0), cmap = 'gray')
```

```python
filter_kernel = np.array([
    [1,1,1],
    [0,0,0],
    [-1,-1,-1]
]) # 'n' - Values of these filters make these learnable using GD - Neural Network
```

```python
output_image = correlate2d(image, filter_kernel, mode = 'same')
plt.imshow(np.maximum(output_image,0), cmap = 'gray')
```

```python
filter_kernel = np.array([
    [-1,0,1],
    [-1,0,1],
    [-1,0,1]
])
output_image = correlate2d(image, filter_kernel, mode = 'same')
plt.imshow(np.maximum(output_image,0), cmap = 'gray')
```

```python
filter_kernel = np.array([
    [1,0,-1],
    [1,0,-1],
    [1,0,-1]
])
output_image = correlate2d(image, filter_kernel, mode = 'same')
plt.imshow(np.maximum(output_image,0), cmap = 'gray')
```

```python
filter_kernel = np.array([
    [-1,-1,-1],
    [1,1,1],
    [0,0,0]
])
output_image = correlate2d(image, filter_kernel, mode = 'same')
plt.imshow(np.maximum(output_image,0), cmap = 'gray')
```
